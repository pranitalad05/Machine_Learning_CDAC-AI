```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from ipywidgets import widget,interact
from sklearn.metrics import root_mean_squared_error
```

```python
# 1. Load Data
df = pd.read_csv(r"D:\Machine_Learning_CDAC-AI\a] Datasets\monthly-milk-production-pounds-p.csv", index_col=0)

# Ensure the index is recognized as a datetime if it isn't already
df.index = pd.to_datetime(df.index)

y_train = df['Milk'].iloc[:-12]
y_test = df['Milk'].iloc[-12:]
```

```python
# 2. Define the Interactive Moving Average Function
def MA(span):
    # Calculate rolling mean on train data
    fcast = y_train.rolling(window=span).mean()
    
    # Get the last calculated rolling average value to project forward
    last_ma_value = fcast.iloc[-1]
    
    # Create the prediction Series matching y_test's length and index
    y_pred = pd.Series(last_ma_value, index=y_test.index)
    
    # Calculate RMSE
    rmse = np.round(root_mean_squared_error(y_test, y_pred), 2)
    
    # Plotting
    plt.figure(figsize=(10, 5))
    plt.plot(y_train.index, y_train, label='Train', c="blue")
    plt.plot(y_test.index, y_test, label='Test', c="orange")
    plt.plot(y_pred.index, y_pred, label=f'MA Forecast (span={span})', c="red", linestyle="--")
    plt.plot(fcast.index, fcast, label='Historical Rolling Mean', c="green", alpha=0.5)
    
    # Dynamically place text based on data instead of hardcoded coordinates
    plt.title(f"Moving Average Forecast | RMSE = {rmse}")
    plt.legend(loc='upper left', frameon=True)
    plt.grid(True, alpha=0.3)
    plt.show()
```

```python
# 3. Trigger the interactive widget
interact(MA, span=(1, 12));
```
